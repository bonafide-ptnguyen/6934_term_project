```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import json

# Load the dataset
file_path = 'datasets/Most_Streamed_Spotify_Songs_2024.csv'
spotify_data = pd.read_csv(file_path, encoding='ISO-8859-1')

# Data Cleaning - Convert stream/count columns to numeric, handling commas and errors
columns_to_numeric = [
    'Spotify Streams', 'Spotify Playlist Count', 'Spotify Playlist Reach',
    'YouTube Views', 'YouTube Likes', 'TikTok Posts', 'TikTok Likes', 'TikTok Views',
    'YouTube Playlist Reach', 'Apple Music Playlist Count', 'AirPlay Spins', 'SiriusXM Spins',
    'Deezer Playlist Count', 'Deezer Playlist Reach', 'Amazon Playlist Count',
    'Pandora Streams', 'Pandora Track Stations', 'Soundcloud Streams', 'Shazam Counts'
]

for col in columns_to_numeric:
    spotify_data[col] = spotify_data[col].astype(str).str.replace(',', '').astype(float)

# Convert 'Explicit Track' to numeric (0 and 1)
spotify_data['Explicit Track'] = pd.to_numeric(spotify_data['Explicit Track'], errors='coerce').fillna(0).astype(int)

print("Data loaded and cleaned successfully!")
```

```python
# Multi-Platform Performance Bar Chart
# Select platforms and metrics for comparison
platforms = ['Spotify Streams', 'YouTube Views', 'TikTok Views']
platform_names = ['Spotify', 'YouTube', 'TikTok'] # for better labels
top_n_songs = 10  # Display top N songs for clarity

# Normalize metrics
normalized_data = pd.DataFrame()
for index, row in spotify_data.head(top_n_songs).iterrows():
    song_name = row['Track']
    song_metrics = {}
    max_val = 0
    for platform_col in platforms:
        value = row[platform_col]
        if not pd.isna(value): # Handle NaN values if any
            max_val = max(max_val, value)

    for i, platform_col in enumerate(platforms):
        value = row[platform_col]
        normalized_value = (value / max_val) if max_val > 0 and not pd.isna(value) else 0
        song_metrics[platform_names[i]] = normalized_value
    normalized_data = pd.concat([normalized_data, pd.DataFrame([song_metrics], index=[song_name])])

# Plot the normalized data
ax = normalized_data.plot(kind='bar', figsize=(12, 6))
plt.title('Normalized Performance Across Platforms (Top {} Songs)'.format(top_n_songs))
plt.ylabel('Normalized Performance Score (0-1)')
plt.xlabel('Song Title')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Correlation Heatmap of Spotify Popularity Factors

# Select numerical columns for correlation analysis
correlation_columns = [
    'Spotify Popularity', 'Spotify Streams', 'Spotify Playlist Count', 'Spotify Playlist Reach',
    'YouTube Views', 'YouTube Likes', 'TikTok Views', 'TikTok Likes', 'TikTok Posts',
    'Apple Music Playlist Count', 'AirPlay Spins', 'SiriusXM Spins',
    'Deezer Playlist Count', 'Deezer Playlist Reach', 'Amazon Playlist Count',
    'Pandora Streams', 'Pandora Track Stations', 'Soundcloud Streams', 'Shazam Counts'
]

correlation_data = spotify_data[correlation_columns].copy()

# Calculate correlation matrix
correlation_matrix = correlation_data.corr(numeric_only=True) # Explicitly set numeric_only

# Create heatmap
plt.figure(figsize=(14, 12))
sns.heatmap(correlation_matrix, annot=False, cmap='coolwarm', fmt=".2f") # Removed annot for better visual clarity, can add annot=True for values
plt.title('Correlation Heatmap of Spotify Popularity Factors')
plt.tight_layout()
plt.show()
```

```python
# Explicit Content Impact Box Plots

# Select metrics for box plot comparison
boxplot_metrics = [
    'Spotify Streams', 'YouTube Views', 'TikTok Views', 'Spotify Playlist Reach'
]
metric_names = ['Spotify Streams (Millions)', 'YouTube Views (Millions)', 'TikTok Views (Billions)', 'Spotify Playlist Reach (Millions)'] # For better labels and scaling

# Create box plots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten() # Flatten axes array for easy iteration

for i, metric in enumerate(boxplot_metrics):
    ax = axes[i]

    # Scale down large numbers for better readability on the plot
    if metric in ['Spotify Streams', 'Spotify Playlist Reach', 'YouTube Views']:
        spotify_data['Scaled_Metric'] = spotify_data[metric] / 1e6  # Scale to millions
    elif metric == 'TikTok Views':
        spotify_data['Scaled_Metric'] = spotify_data[metric] / 1e9  # Scale to billions
    else:
        spotify_data['Scaled_Metric'] = spotify_data[metric] # No scaling for others if needed

    sns.boxplot(y='Scaled_Metric', x='Explicit Track', hue='Explicit Track', data=spotify_data, ax=ax, palette=['lightgreen', 'lightcoral'], dodge=False) # Vertical boxplots, hue='Explicit Track', dodge=False to remove margin
    ax.set_title('{} Distribution by Explicit Content'.format(metric_names[i]))
    ax.set_xlabel('Explicit Track (0: No, 1: Yes)') # x-axis label is Explicit Track
    ax.set_ylabel(metric_names[i]) # y-axis label is the metric

    spotify_data.drop(columns=['Scaled_Metric'], inplace=True) # Clean up temporary scaled column

plt.tight_layout()
plt.show()
```

### 5. "How" Analysis: Design of Visuals

#### (Design 1) Choropleth Map

* **Connection to What/Why Target Pairs:**
    * **What:** Visualizes the global distribution of average daily rank for a selection of top Spotify songs.
    * **Why:** Addresses **Research Question 1**: To understand the **Global Reach and Popularity Distribution** of top songs and identify **geographical clusters or outliers** in their success.
* **Idioms Involved:**
    * **Choropleth Map:** Uses color intensity applied to country polygons to represent the average daily rank of the selected top songs within each country.
* **Methods Used:**
    * **Data Integration:** Combines the "Top Spotify Songs in 73 Countries" dataset with the world GeoJSON file.
    * **Filtering:** Filters the "Top Spotify Songs in 73 Countries" dataset for a specific time period (e.g., a year) and potentially for a pre-defined set of top global songs.
    * **Aggregation:** Calculates the average daily rank for the selected songs within each country present in the dataset. This involves grouping the data by country and then calculating the mean of the `daily_rank` for the songs of interest.
    * **Geographical Encoding:** Maps the calculated average daily rank for each country to the corresponding country polygon in the GeoJSON data using country codes or names as identifiers.
    * **Color Encoding:** Employs a sequential color scale where the intensity of the color represents the average daily rank. #### (Design 2) Linked Bar Charts

* **Connection to What/Why Target Pairs:**
    * **What:** Presents two linked bar charts:
        * One showing the frequency of appearance of top tracks within each of the 73 countries.
        * Another showing a selected characteristic (e.g., average popularity, explicit content prevalence) of those frequently appearing top tracks.
    * **Why:** Addresses **Research Question 2**: To identify **Country-Specific Top Tracks** and explore their **Shared Characteristics** and potential **regional patterns** in these characteristics.
* **Idioms Involved:**
    * **Multiple Bar Charts:** Two bar charts are displayed side-by-side or stacked.
    * **Linking (Highlighting):** Selecting a country in the first bar chart (showing top track frequency) highlights the corresponding data points (the frequently appearing tracks and their characteristics) in the second bar chart.
* **Methods Used:**
    * **Data Aggregation (Chart 1):** For each of the 73 countries, the code counts the occurrences of each unique track within the top charts over the observed period. The first bar chart displays countries on one axis and the count (frequency) of top tracks on the other.
    * **Data Aggregation (Chart 2):** For the frequently appearing top tracks identified in each country, the code calculates or retrieves the value of a selected characteristic (e.g., average popularity from the original dataset, or an indicator for explicit content). The second bar chart displays these tracks (or potentially aggregated by artist or a broader category) on one axis and the value of the chosen characteristic on the other.
    * **Linking Mechanism:** An interactive element (e.g., using a library like Plotly with `callback` functions or similar) connects the two charts. When a bar representing a country is selected in the first chart, the second chart dynamically updates or highlights the bars corresponding to the top tracks from that specific country and their associated characteristics. This allows for direct comparison of track characteristics for top songs within a selected country.

**Implementation:**

```python
if spotify_data is not None and universal_spotify_data is not None and countries_geo is not None:
    print("Spotify data loaded successfully:", spotify_data.head())
    print("Universal Spotify data loaded successfully:", universal_spotify_data.head())
    print("Countries GeoJSON loaded successfully.")

    # Filter the universal Spotify data for the desired date range (Jan-01 to Dec-31)
    start_date = '2024-01-01'
    end_date = '2024-12-31'
    universal_spotify_data['snapshot_date'] = pd.to_datetime(universal_spotify_data['snapshot_date'])
    filtered_universal_data = universal_spotify_data[universal_spotify_data['snapshot_date'].between(start_date, end_date)]

    # Merge the datasets to get track names
    merged_data = pd.merge(filtered_universal_data, spotify_data[['Track', 'ISRC']], left_on='name', right_on='Track', how='inner')

    # Define the countries for which you want to see the top tracks
    selected_countries = ['US', 'GB', 'CA', 'AU', 'DE']

    # Filter the merged data for the selected countries
    country_specific_data = merged_data[merged_data['country'].isin(selected_countries)]

    # Group by country and track name, then count the occurrences (appearances in the top charts)
    top_tracks_country = country_specific_data.groupby(['country', 'name']).size().reset_index(name='count')

    # Sort the tracks within each country by their count in descending order
    top_tracks_country_sorted = top_tracks_country.sort_values(['country', 'count'], ascending=[True, False])

    # Get the top 10 tracks for each country
    top_n = 10
    top_tracks_country_filtered = top_tracks_country_sorted.groupby('country').head(top_n).reset_index(drop=True)

    # Create the linked bar charts using facets
    fig = px.bar(top_tracks_country_filtered,
                 y='name',
                 x='count',
                 color='country',
                 facet_col='country',
                 facet_col_wrap=3,
                 orientation='h',
                 height=300 * len(selected_countries) // 3, # Adjust height based on the number of countries
                 title=f'Top {top_n} Tracks in Selected Countries (Jan-Dec 2024)',
                 labels={'name': 'Track Name', 'count': 'Number of Appearances in Top Charts', 'country': 'Country'})

    # Update layout for better readability
    fig.update_layout(yaxis={'categoryorder': 'total ascending'}) # Sort tracks by count within each facet
    fig.for_each_annotation(lambda a: a.update(text=a.text.split("=")[1])) # Remove country= prefix from facet titles

    fig.show()

else:
    print("Could not proceed with visualization due to missing datasets.")
```

**Critique:**

* **Strengths:** Powerful for exploring potential correlations or relationships between different variables. Interactive exploration allows for detailed investigation of specific data points. Facilitates comparative analysis of different metrics.
* **Weaknesses:** Can become complex and overwhelming with too many categories or characteristics. Requires user interaction to realize its full potential. Interpretation depends on the clarity of the linking mechanism.

#### (Design 3) Heatmap with Interactive Features

* **Connection to What/Why Target Pairs:**
    * **What:** Displays a heatmap visualizing the average daily rank of a selection of individual songs across the 73 countries.

* **Idioms Involved:**
    * **Heatmap:** Uses a grid of colored cells where each cell represents the average daily rank of a specific song in a specific country. The color intensity corresponds to the rank value.
* **Methods Used:**
    * **Data Filtering:** Allows the analyst to select a specific song from the dataset (potentially through a dropdown menu or other interactive control).
    * **Data Aggregation:** Once a song is selected, the code filters the "Top Spotify Songs in 73 Countries" dataset to include only the data for that specific song. Then, it calculates the average daily rank of that song for each country where it appeared in the charts.
    * **Matrix Representation:** Organizes the aggregated data into a matrix where rows represent the selected song (or potentially a small set of pre-selected songs for comparison), columns represent the 73 countries, and the cell values are the calculated average daily ranks.
    * **Color Encoding:** Applies a color scale to the heatmap cells, where the color intensity represents the average daily rank. A reversed scale is likely used, with brighter colors indicating better performance (lower average rank) and darker colors indicating worse performance.
    * **Interactive Features:**
        * **Song Selection:** An interactive element (e.g., a dropdown) allows the user to choose which song's global performance they want to visualize.
        * **Hover Tooltips:** When the user hovers over a cell, a tooltip appears displaying the song name, the country, and the specific average daily rank value for that song in that country.
        * **Potential Filtering/Highlighting:** Additional interactive controls could allow filtering or highlighting countries based on specific rank ranges or other criteria for the selected song.

**Implementation:**


```python
if spotify_data is not None and universal_spotify_data is not None and countries_geo is not None:
    print("Spotify data loaded successfully:", spotify_data.head())
    print("Universal Spotify data loaded successfully:", universal_spotify_data.head())
    print("Countries GeoJSON loaded successfully.")

    # Filter the universal Spotify data for the desired date range (Jan-01 to Dec-31)
    start_date = '2024-01-01'
    end_date = '2024-12-31'
    universal_spotify_data['snapshot_date'] = pd.to_datetime(universal_spotify_data['snapshot_date'])
    filtered_universal_data = universal_spotify_data[universal_spotify_data['snapshot_date'].between(start_date, end_date)]

    # Select the top N global songs based on 'All Time Rank' and get their track names
    top_n_global = 20
    top_global_tracks = spotify_data.sort_values('All Time Rank').head(top_n_global)['Track'].tolist()
    print(f"Top Global Tracks: {top_global_tracks}")

    # Filter the universal data to include only the top global tracks
    song_performance_filtered = filtered_universal_data[filtered_universal_data['name'].isin(top_global_tracks)]
    print(f"Shape of song_performance_filtered: {song_performance_filtered.shape}")

    # Calculate the average daily rank for each song in each country
    average_rank = song_performance_filtered.groupby(['name', 'country'])['daily_rank'].mean().reset_index()

    # Pivot the table to have countries as columns and songs as rows
    pivot_rank = average_rank.pivot(index='name', columns='country', values='daily_rank')

    # Identify countries with data for these top songs
    countries_in_data = [col for col in pivot_rank.columns if col in filtered_universal_data['country'].unique()]

    # Filter out songs that have NaN for all countries
    pivot_rank_with_data = pivot_rank[countries_in_data].dropna(axis=0, how='all')
    print(f"Number of songs in pivot_rank_with_data: {pivot_rank_with_data.shape[0]}")
    print(f"Songs in pivot_rank_with_data: {pivot_rank_with_data.index.tolist()}")

    # Create the heatmap
    fig = px.imshow(pivot_rank_with_data,
                    labels=dict(x="Country", y="Track", color="Average Daily Rank"),
                    x=pivot_rank_with_data.columns,
                    y=pivot_rank_with_data.index,
                    color_continuous_scale=px.colors.sequential.Viridis[::-1], # Use a reversed scale so lower rank (better) is a brighter color
                    title=f'Average Daily Rank of Top {top_n_global} Songs Across Countries (Jan-Dec 2024)',
                    aspect="auto") # Adjust aspect ratio to prevent squishing

    fig.update_layout(height=800) # Increase height to accommodate more tracks

    fig.show()

else:
    print("Could not proceed with visualization due to missing datasets.")
```
